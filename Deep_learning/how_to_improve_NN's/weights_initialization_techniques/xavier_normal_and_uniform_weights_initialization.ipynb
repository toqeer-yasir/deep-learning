```python
import numpy as np
from sklearn.datasets import make_blobs
from mlxtend.plotting import plot_decision_regions
from tensorflow import keras
from tensorflow.keras.layers import Input, Dense
```

```python
X , y = make_blobs(n_samples= 100, n_features=2, cluster_std=2.5, centers=2, random_state=44)
```

```python
import matplotlib.pyplot as plt
plt.style.use('default')
```

```python
plt.scatter(X[:, 0], X[:, 1], c=y)
```

```python
from tensorflow.keras import Sequential
```

```python
model = Sequential()
```

```python
model.add(Input(shape= (2, )))
model.add(Dense(10, activation='tanh', kernel_initializer='glorot_normal'))
model.add(Dense(10, activation='tanh', kernel_initializer='glorot_normal'))
model.add(Dense(10, activation='tanh', kernel_initializer='glorot_normal'))
model.add(Dense(10, activation='tanh', kernel_initializer='glorot_normal'))
model.add(Dense(1, activation='sigmoid'))
```

glorot_normal is also called xavier_normal and its formula for weights initialization is: np.sqrt(1/fan_in), fan_in is the total numbers of inputs.

glorot_uniform is called xavier_uniform and it's weights are initialized in the range of this output : range = np.sqrt(6/fan_in + fan_out) [- range, + range] fan_out is the total number of inputs out from that neuron.

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
model.fit(X, y, epochs=100, validation_split=0.2, verbose=1)
```

```python
plot_decision_regions(X, y, clf=model)
```
