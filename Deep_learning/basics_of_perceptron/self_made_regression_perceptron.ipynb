```python
import numpy as np
import pandas as pd
```

```python
class Perceptron():
    def __init__(self, lr: float = 0.01, max_iter: int = 1000, random_state=None):
        self.weights = None
        self.bias = None
        self.lr = lr
        self.max_iter = max_iter
        self.coef_ = None
        self.intercept_ = None
        if random_state is not None:
            np.random.seed(random_state)

    def loss_function(self, y_true, y_pred):
        """mse loss function"""
        return np.mean((y_pred - y_true) ** 2)

    def fit(self, X, y):
        """training through gradient descent"""
        X = np.array(X)
        y = np.array(y).reshape(-1, 1)
        
        n_samples, n_features = X.shape
        
        self.weights = np.random.randn(n_features, 1) * 0.01
        self.bias = 0.0
        
        for epoch in range(self.max_iter):
            y_pred = np.dot(X, self.weights) + self.bias
            
            error = y_pred - y
            dw = (1 / n_samples) * np.dot(X.T, error)  #  1/n_samples for mse normalization
            db = (1 / n_samples) * np.sum(error)       # np.sum() to get scalar
            
            self.weights -= self.lr * dw
            self.bias -= self.lr * db
            
        self.coef_ = self.weights.flatten() # flatten() convers negative to a whole positive number
        self.intercept_ = self.bias
        
        return self

    def predict(self, X):
        """Predict using the trained model"""
        X = np.array(X)
        y_pred = np.dot(X, self.weights) + self.bias
        return y_pred.flatten()
```

```python
df = pd.read_csv('salary.csv')
```

```python
df.head()
```

```python
x, y = df.iloc[:, :-1], df['salary_in_pkr']
```

```python
x.head()
```

```python
y.head()
```

```python
pr = Perceptron()
```

```python
x, y = np.array(x), np.array(y)
```

```python
pr.fit(x, y)
```

```python
print(pr.coef_, pr.intercept_)
```

```python
result = pr.predict(x)
print('Actual_label vs Predicted_label:\n')

total = 0
for actual, predicted in zip(y, result):
    if actual == predicted:
        total += 1
        print(f'Correct: {actual:.0f} = {predicted:.0f}')
    else:
        print(f'Wrong: {actual:.0f} != {predicted:.0f}')
print(f"\nAccuracy: {total/len(result)}")
```
