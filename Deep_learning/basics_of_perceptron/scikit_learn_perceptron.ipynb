#### a perceptron is the basic unit of deep-learning it takes all the values of x and the output of each data-point then auto calculate weight for each x to decide which x has how much importance(weight) in predicting the value of y. a single perceptron draws a linear-line to classify into different regions on the basis of y.(in simple words it is a binary classifier). If there are 3 x(features) it draws a hyper-plain.

## how to use perceptron.(basic)

```python
import numpy as np, pandas as pd
from  matplotlib import pyplot as plt
import seaborn as sns
from sklearn.linear_model import Perceptron
from mlxtend.plotting import plot_decision_regions
```

```python
df = pd.read_csv('placement.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
X = df.iloc[:, :-1]
X.head()
```

```python
y = df['placed']
y.shape
```

```python
sns.scatterplot(x= df['cgpa'], y= df['resume_score'], hue= df['placed'], palette= ['red', 'green'])
```

```python
pt = Perceptron(penalty='l1', alpha=0.1, max_iter=500)
```

```python
pt.fit(X.values, y.values)
```

```python
pt.coef_
```

```python
pt.intercept_
```

```python
plot_decision_regions(X.values, y.values, clf= pt, legend= 2)
```

## Now with ml traditional algorithm

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression()
```

```python
lr.fit(X.values, y.values)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
plot_decision_regions(X.values, y.values, clf= lr, legend= 2)
```

### we can clearly see ml's logistic regression performed better because single perceptron is not enough with very less data its accuracy improves with more data.
